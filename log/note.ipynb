```python
import datetime

def record(note):
    now = datetime.datetime.now().strftime("%Y-%m-%d %H:%M:%S")
    print(f"[{now}]\n{note.strip()}")

record("""
This is a test message.
It captures the current timestamp and formats it along with the provided message.
""")
```

```python
record("""
This is another test message.
It demonstrates how to use the record function to log messages with timestamps.
""")
```

```python
record("""
Functions are classes. Classes are functions.
There are three types of functions:
1. Ordinary function, which is like a function in C
2. Passive function, which is like a struct in C
3. Active function, which is like a class in C
""")
```
