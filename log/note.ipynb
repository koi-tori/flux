```python
import datetime

def record(note):
    now = datetime.datetime.now().strftime("%Y-%m-%d %H:%M:%S")
    print(f"[{now}]\n{note.strip()}")

record("""
This is a test message.
It captures the current timestamp and formats it along with the provided message.
""")
```

```python
record("""
Functions are classes. Classes are functions.
There are three types of functions:
1. Ordinary function, which behaves like functions in other language
2. Passive function, which is used to store data
3. Active function, which manages data
If a function is passive or active, its context is reserved until it's deleted explicitly or implicitly when the scope ends. It's like a portable toolkit or drawer for us to use. Hence, we call it a "drawer function".
A passive function doesn't contain complex logics, its size and structure should be inferred from defination.
You can use "." to easily get access to its members. Also, there is a syntactic sugar:
func.draw:
    # You can manipulate data here. It should be clear that the current scope is completely inside the function. This means you should use "fetch nonlocal" to get names outside the code block, and .
An active function, which is like a "thread", uses the key word "yield" to wait for yield in its body. The owner of the function can use the statement "resume func" to make the function continue.
For active functions, there are also some sugars to eat:
func.accept:
    # This means "draw, close and let wait", often used to deliver parameters to the function.
func.yield:
    # This means "draw, close and run", often used to get a yield.
""")
```

```python
record("""
A continuous memory is called a "structure", which is written between "[" and "]".
You can define a structure in a convenient way:
var x = [
    arr = [1, 2, 3],
    obj = [ a = 1, b = 2 ]
]
A structure that contains the same type of data is called an "array".
var arr = [first = 1, 2, 3]
Use "type" statements to define the type of data in a structure:
typedef MyStruct(int _x) = [
    x = _x,
    data = [1, 2, 3],
    info = [ a = 1, b = 2 ]
]
""")
```
